# NPUWattch hands-on — 3. Timeloop: an Eyeriss-like accelerator running AlexNet

This is the basic Timeloop example. The design is an Eyeriss-like accelerator
with a 14×12 array of 168 MAC units (one int8 MAC per PE, 16-bit accumulate), a 128 KB global buffer, and LPDDR4 DRAM. The
workload is all eight layers of AlexNet, one `timeloop-mapper` stats file per
layer.

The example folder has:

```
timeloop/eyeriss_like/
├── arch.yaml                   the Accelergy v0.4 architecture, the same file Timeloop reads
├── compound_components.yaml    NPUWattch definition file 1: a class made of several primitives
├── projection.yaml             NPUWattch definition file 2: Timeloop events -> compound elements
├── user_components.yaml        NPUWattch definition file 3: cost of custom blocks
└── alexnet/
    ├── stats/01_conv1.stats.txt ... 08_fc8.stats.txt
    ├── run                     the command
    └── out/                    report.html, report.json (written by ./run)
```

The command in `run` is:

```bash
npuwattch --harness timeloop --arch-yaml ../arch.yaml --stats stats/ \
          --node 7nm --clock-mhz 1000 --tree --report out/
```


````python
import sys; sys.path.insert(0, ".")
from tutorial_helpers import *
import pandas as pd
pd.set_option("display.width", 160); pd.set_option("display.max_colwidth", 60)
print("npuwattch:", npuwattch_binary())
````

## 1. Run it, and read the tree first

The run takes a few seconds. The first thing to look at is the instance tree:
the hardware as NPUWattch understood it from `arch.yaml`.

````python
ex = example("timeloop/eyeriss_like/alexnet")
text = run(ex)
show_tree(text)
machine_card(report(ex))
````

Three things to check in the tree:

- **Six components**: `DRAM`, `shared_glb`, three per-PE scratchpads, and `mac`.
  They are the six `!Component` nodes of `arch.yaml`.
- **`[×168]`** on the scratchpads and the MAC. The `!Container` nodes
  `PE_column` (`meshX: 14`) and `PE` (`meshY: 12`) above them multiply out to
  168 instances.
- **The sizes** at the end of each line: `shared_glb` is 128 KB, the weight
  scratchpad is 384 B per PE. If a size is wrong here, every number after it
  is wrong too, so this is the place to catch a mistake in the architecture
  file.

The `Machine:` line under the tree adds the clock. Timeloop stats have no
clock, so the 1 GHz comes from `--clock-mhz` in the `run` script. It turns
the cycles into time, and the leakage power into leakage energy. It does not
change the cycle counts or the dynamic energy.

## 2. The messages

Every assumption NPUWattch makes is printed. This run has no WARNING lines.
The INFO lines tell you what was taken from a convention rather than from your
file. (The PyTorchSim run in notebook 1 had WARNING lines as well. The
difference is that this architecture file states every size NPUWattch needs;
where a port count is missing, the INFO line `NW-7307` names the default.)

````python
show_messages(text, kinds=("WARNING", "INFO"))
````

What they mean:

- `NW-7306`, `depth 16384 is the total of 32 banks, so mem_depth_per_bank is
  512`: Accelergy's `depth` is the whole buffer; NPUWattch's SRAM model works per
  bank. The harness converted it and tells you the result (128 KB in total).
- `NW-7307`, `no port count is declared, so NPUWattch uses one shared
  read-write port`: the buffer has no `n_rw_ports` in `arch.yaml`, so one port is assumed, and the
  port count is checked against the bandwidth the mapping needs.
- `NW-7301`, `the Accelergy attribute(s) datawidth have no effect in
  NPUWattch`: NPUWattch reads the word width from `width`; `datawidth` has
  no NPUWattch attribute with the same meaning, so it changes nothing.
- `NW-7403`, `8 pJ/bit from the shipped table lpddr4.yml`: the DRAM cost is
  the 8 pJ/bit of the shipped LPDDR4 table. It is an all-in value, so the
  activation and refresh rows of the DRAM breakdown (`NW-1040`) are 0 by
  design. Notebook 5 swaps in a different table.
- `NW-2116`, `User component ... is in the library, but the run does not use
  it`, and `NW-7101`, `Compound component 'counter' ... is parsed, but no
  component uses it`: the definition files contain an example compound
  (`counter`) and an example user component that this design does not use.
  Nothing to do.
- `NW-3107`, `Clock gating is assumed: idle cycles of 4 logic component(s)
  cost leakage only.`: an idle MAC or scratchpad costs leakage only, and each
  access costs its per-access energy. This line is in every run.
- `NW-1902`, `Message summary: 0 CRITICAL, 0 ERROR, 0 WARNING, 26 INFO.`: the
  last message of every run.

## 3. One window per layer

`--stats stats/` is a directory, so each stats file becomes one **window** in
the report, in file-name order. That is why the files are named `01_`...`08_`.

````python
rep = report(ex)
w = windows_df(rep)
display(w.drop(columns="total_pJ"))
ax = w.plot.bar(x="window", y="total_pJ", legend=False, figsize=(8, 3), title="Energy per layer (pJ)")
ax.set_xlabel(""); ax.set_ylabel("pJ");
````

The fully connected layers dominate the total, although they do little
compute: fc6 and fc7 alone are 3.5 of the 5.3 mJ. Look at the pJ / MAC.
The table below divides each window by the MACs that Timeloop counted for the
layer (`Computes (total)` in the stats file).

````python
import re
def layer_macs(example_dir):
    macs = {}
    for f in sorted((example_dir / "stats").glob("*.stats.txt")):
        m = re.search(r"Computes \(total\)\s*:\s*(\d+)", f.read_text())
        macs[f.name.replace(".stats.txt", "")] = int(m.group(1))
    return macs

def window_components(rep, label, top=8):
    """The per-component energy of one window, from report.json."""
    labels = [w["label"] for w in rep["windows"]]
    j = labels.index(label)
    rows = [{"component": r["name"].replace("system_top_level.", ""), "energy": r["cells"][j]}
            for r in rep["matrix"]["rows"] if r["cells"][j] != "—"]
    return pd.DataFrame(rows).head(top)

macs = layer_macs(ex)
w["MACs"] = [macs[k] for k in w["window"]]
w["pJ / MAC"] = (w["total_pJ"] / w["MACs"]).round(1)
w["MAC util %"] = (100 * w["MACs"] / (w["cycles"] * mac_units(rep))).round(1)   # 168 MAC units
display(w.drop(columns="total_pJ"))
````

**The MAC utilization column.** It is the MACs of the layer divided by the
cycles times the 168 MAC units. The convolution layers are at 57 to 93%: the
mapping keeps most of the 14×12 array busy. The fully connected layers are
at 10%: the array waits for weights from DRAM. Read the `pJ / MAC` column with
this column next to it.

**Why the convolution layers cost 1.4 to 3.8 pJ per MAC and the fully
connected layers 65.** A convolution reuses every weight at every output
position (55×55 for conv1, 13×13 for conv5), so one weight read from DRAM
serves many MACs from the on-chip scratchpads.

conv2 is the exception. Its mapping reads each weight 27 times from DRAM
(8.3 M reads for 307 k weights in the stats file) and spills partial sums
to DRAM and back. That is why conv2 costs 3.8 pJ per MAC and the other
convolutions 1.4 to 1.7.

A fully connected layer with batch 1 uses each weight exactly once: fc6 reads its 9216×4096 int8
weights, 37.7 MB, from DRAM for 37.7 M MACs. At 8 pJ/bit that is 64 pJ per
weight byte, and 37.7 M × 64 pJ = 2.4 mJ, which is the DRAM line of the fc6
window below. Nothing on chip can help: there is no reuse to capture.

## 4. Where the energy went

````python
display(components_df(rep))
show_totals(text)
````

Expected result at 7 nm, 1 GHz:

| | |
| --- | --- |
| Total energy | about 5.3 mJ for the eight layers (7.4 pJ per MAC) |
| Area | 0.459 mm² |
| On-chip energy per MAC | about 1.0 pJ |
| DRAM share | about 87% |

The `model` column says where each number came from: `cal` is a trained model,
`const` is a cost table (the DRAM). Every row here is one or the other.

**Why the MAC is the largest on-chip consumer.** Eyeriss is a row-stationary
design: each PE keeps a row of weights in a 384 B scratchpad and reads its
weights from there, not from the global buffer. That scratchpad is a
192-entry register file. With clock gating, an access costs only its
per-access energy: 0.18 pJ for a read and 0.53 pJ for a write (the per-access
energies of the report), against 0.66 pJ for the int8 MAC. A weight read costs
0.27 times a MAC. So the `mac` row is 8.8% of the energy and the
`weights_spad` row 2.0%. The input and partial-sum scratchpads are smaller
(12 and 16 entries), and their accesses cost 0.03 to 0.09 pJ. The three
scratchpads together are about 4% of the energy.

**Why the global buffer is almost free.** The 128 KB `shared_glb` is read
once per PE row, not per MAC: the row-stationary dataflow multicasts a row
of inputs to a diagonal of PEs. Its 0.3% share is the result of this dataflow.

**Why DRAM is 87% although the convolutions reuse everything.** Two of the
eight layers (fc6, fc7) have no reuse, and they alone are 3.5 of the 4.6 mJ
of DRAM energy. The per-window table below shows it: in conv2, DRAM leads
(633 µJ), then the MAC (147 µJ), then the partial-sum scratchpad (31 µJ); in
fc6, DRAM is almost all of the energy.

````python
display(window_components(rep, "02_conv2"))
display(window_components(rep, "06_fc6"))
````

**What to look at.** On this run, the fully connected layers and DRAM are
where the energy is: fc6 and fc7 read each weight from DRAM for one MAC. This
run has batch 1; a larger batch would reuse each weight and move these
layers. On chip, the int8 MAC is the largest consumer. The register
files that feed it are cheap when they are clock-gated: the three
scratchpads together cost less than half of the MAC.

## 6. The HTML report

`out/report.html` is a single self-contained page. From top to bottom: the
totals, the energy and area breakdown charts, the DRAM breakdown, the energy
across windows, the component table, the instance tree, and a provenance
section that lists every input file and every message of the run. The same
numbers are in `out/report.json` for scripts (that is what the tables above
came from).

**The average-power line.** In the chart of the energy across windows, the
line is the average power of each window: its energy divided by its time (the cycles at
1 GHz).
The bar width is proportional to the cycles. The fully connected layers have
no reuse, so almost all of their energy is DRAM energy, and the DRAM
bandwidth of 16 B per cycle sets their cycles. Thus their power is higher
than the power of the convolution layers (about 1 W against 0.15 to 0.36 W):
the fc6 bar is about as wide as the conv2 bar, but almost three times as
tall.

## Messages you will see, and what to do

**Every time.** The INFO lines in section 2, including `NW-3107`, which says
that NPUWattch assumes clock gating. None of them needs action.

**If something looks wrong:**

| What you see | What it means | What to do |
| --- | --- | --- |
| `WARNING (NW-7219): Stats level(s) with no matching description component: X` | A level in the stats file has a name that is not a component name | Add a `stats_map.yaml` with `levels: {X: <component>}` and pass `--stats-map`, or list it under `ignore:` |
| `WARNING (NW-7114): N component(s) have no NPUWattch primitive and are not modeled: ...` | An Accelergy `class` NPUWattch does not know | Define the class as a compound in `compound_components.yaml`, or add it to `user_components.yaml` |
| `ERROR (NW-5101): projection.yaml is not in <dir>, and --projection is not given.` | A definition file is missing | Copy the three files from `timeloop/eyeriss_like/` next to your `arch.yaml`, or pass `--projection` |
| `WARNING (NW-7309): ... bandwidth N words/cycle needs M accesses/cycle, but ... give ...` | The mapping needs more ports than the buffer declares | Add `n_rd_ports` / `n_wr_ports` to the buffer in `arch.yaml`, as `psum_spad` does here |
| The power of a layer with no reuse is several times too high | `arch.yaml` gives the DRAM no bandwidth, so Timeloop counts the cycles from the PEs alone | Add `read_bandwidth` / `write_bandwidth` to the `DRAM` component, as this example does |
| `npuwattch: command not found` | NPUWattch is not installed in this environment | `cd NPUWattch && pip install -e .` |
