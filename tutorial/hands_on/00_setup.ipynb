## How the notebooks work

Each run is one call to `run(example_dir, ...)`. It executes the example's
`./run` script, which is the same `npuwattch` command you would type in a
terminal, and returns the console text. The full text is saved to
`out/console.txt` in the example folder, so you can always open it. The
notebooks then show one part of that text at a time: the instance tree, the
messages, the totals. The tables come from `out/report.json`, and the HTML
report is shown inside the notebook.

If a cell fails with `npuwattch is not on your PATH`, install NPUWattch (see
above) and restart the kernel.
